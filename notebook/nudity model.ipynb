```python
from PIL import Image
img1 = Image.open('data/test_image1.jpg')
```

```python
from torchvision import transforms

def preprocess(img1):
    img1 = img1.resize((224, 224))
    convert_tensor = transforms.ToTensor()
    t1 = convert_tensor(img1)
    m =  torch.tensor([0.485, 0.456, 0.406])
    std = torch.tensor([0.229, 0.224, 0.225])
    t1 = (t1 - m.reshape(-1, 1, 1))/ std.reshape(-1, 1, 1)
    return torch.unsqueeze(t1, axis = 0)
```

```python
model(preprocess(img1))
```

```python
model(preprocess(Image.open('data/test_image2.png').convert('RGB')))
```

```python
model(preprocess(Image.open('data/test_image3.png').convert('RGB')))
```
