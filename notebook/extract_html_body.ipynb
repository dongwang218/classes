```python
from bs4 import BeautifulSoup
def parse_conversation(html, node_selector, user_selector, post_selector):
    soup = BeautifulSoup(html, 'html.parser')
    li_nodes = soup.select(node_selector)
    conversations = []
    
    for node in li_nodes:
        user_node = node.select_one(user_selector)
        post_node = node.select_one(post_selector)
        
        if user_node and post_node:
            conversation = {
                'user': user_node.text.strip(),
                'post': post_node.text.strip()
            }
            conversations.append(conversation)
    
    return conversations
            
```

```python
from selectolax.parser import HTMLParser
def extract_text_with_spacing(html):
    # Parse the HTML content
    parser = HTMLParser(html)
    
    # Tags that might need extra space before or after
    tags_to_modify = ['a', 'br', 'strong', 'em', 'span', 'p', 'div']
    
    for tag in tags_to_modify:
        for node in parser.tags(tag):
            # Add space before and after the tag
            node.replace_with(f' {node.html} ')
    
    # Extract text with spaces handled
    text = parser.body.text(separator=' ')
    
    # Normalize multiple spaces to a single space
    normalized_text = ' '.join(text.split())
    
    return normalized_text
# Example HTML content
html_content = '''
<div class="inner" data-msgid="797" id="msg_797">
    If you are in UK, take a look at <a href="http://sbmcc.co.uk" class="bbc_link" target="_blank" rel="noopener">http://sbmcc.co.uk</a>.<br>You can register as a Forum Guest for free. Tons of info!
</div>
'''
# Extract text with proper spacing
extracted_text = extract_text_with_spacing(html_content)
print(extracted_text)
```
